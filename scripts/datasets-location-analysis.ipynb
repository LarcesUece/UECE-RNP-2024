```
import math
import pandas as pd
import numpy as np
```

Simplified coordinate system for the 26 states of Brazil (including the Federal District). The coordinates are adjusted to reflect the relative positions and distances of the states:

```
brazil_states_coordinates = {
    "ac": (0, 6),   # Acre
    "al": (16, 2),  # Alagoas
    "ap": (10, 16), # Amapá
    "am": (2, 10),  # Amazonas
    "ba": (14, 4),  # Bahia
    "ce": (18, 8),  # Ceará
    "df": (12, 6),  # Distrito Federal
    "es": (16, 1),  # Espírito Santo
    "go": (10, 5),  # Goiás
    "ma": (15, 10), # Maranhão
    "mt": (6, 6),   # Mato Grosso
    "ms": (4, 4),   # Mato Grosso do Sul
    "mg": (14, 3),  # Minas Gerais
    "pa": (8, 12),  # Pará
    "pb": (19, 5),  # Paraíba
    "pr": (6, 1),   # Paraná
    "pe": (18, 4),  # Pernambuco
    "pi": (16, 8),  # Piauí
    "rj": (15, 1),  # Rio de Janeiro
    "rn": (19, 6),  # Rio Grande do Norte
    "rs": (5, 0),   # Rio Grande do Sul
    "ro": (1, 5),   # Rondônia
    "rr": (3, 14),  # Roraima
    "sc": (6, 0),   # Santa Catarina
    "sp": (10, 2),  # São Paulo
    "se": (15, 3),  # Sergipe
    "to": (12, 8),  # Tocantins
}
```

```
data = pd.read_csv("../results/bi-lstm/evaluation_rmse_mae.csv")

# Function to calculate Euclidean distance
def calculate_distance(link):
    city1, city2 = link.split("-")
    coord1 = brazil_states_coordinates.get(city1)
    coord2 = brazil_states_coordinates.get(city2)
    if coord1 and coord2:
        return math.sqrt((coord2[0] - coord1[0])**2 + (coord2[1] - coord1[1])**2)
    return None  # Return None if coordinates are missing

# Add a 'distance' column
data['distance'] = data['link'].apply(calculate_distance)

# Save the updated dataset to a new file
data.to_csv("../results/bi-lstm/evaluation_rmse_mae_distance.csv", index=False)

print(data.head())
```

```
data = pd.read_csv("../results/bi-lstm/evaluation_rmse_mae_distance.csv")

# Calculate the correlation between 'RMSE' and 'distance'
correlation = data['RMSE'].corr(data['distance'])

print(f"Correlation between RMSE and distance: {correlation}")
```

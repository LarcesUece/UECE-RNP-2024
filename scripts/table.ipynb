```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Ler os dados
data = pd.read_csv('regression_Vazao_bbr_rmse.csv')

sns.set_palette("husl")

def plot_model_comparison(data):
    """
    Gera um gráfico de barras comparando o erro médio de cada modelo
    """
    # Calcular a média dos erros por modelo, excluindo a coluna 'source'
    model_means = data.drop('source', axis=1).mean().sort_values()
    
    # Criar figura
    plt.figure(figsize=(15, 8))
    bars = plt.bar(range(len(model_means)), model_means.values)
    
    # Customizar o gráfico
    plt.title('Comparação do Erro Médio entre Modelos de Regressão', pad=20, size=14)
    plt.xlabel('Modelos', labelpad=10)
    plt.ylabel('Erro Médio', labelpad=10)
    
    # Rotacionar e ajustar labels do eixo x
    plt.xticks(range(len(model_means)), model_means.index, rotation=45, ha='right')
    
    # Adicionar valores sobre as barras
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height,
                f'{height:.4f}',
                ha='center', va='bottom')
    
    plt.tight_layout()
    plt.savefig('model_comparison.png', dpi=300, bbox_inches='tight')
    plt.close()

def plot_state_heatmap(data):
    """
    Gera um heatmap mostrando o erro por estado e modelo
    """
    # Selecionar os 5 estados com maior média de erro
    state_means = data.groupby('source').mean().mean(axis=1)
    top_states = state_means.nlargest(5).index
    
    # Filtrar dados para os top 5 estados
    heatmap_data = data[data['source'].isin(top_states)].set_index('source')
    
    # Criar figura
    plt.figure(figsize=(15, 8))
    
    # Criar heatmap
    sns.heatmap(heatmap_data, 
                annot=True, 
                fmt='.4f',
                cmap='YlOrRd',
                center=heatmap_data.mean().mean(),
                cbar_kws={'label': 'Erro'})
    
    plt.title('Heatmap de Erro por Estado e Modelo', pad=20, size=14)
    plt.xlabel('Modelos', labelpad=10)
    plt.ylabel('Estados', labelpad=10)
    
    # Rotacionar labels do eixo x
    plt.xticks(rotation=45, ha='right')
    
    plt.tight_layout()
    plt.savefig('state_heatmap.png', dpi=300, bbox_inches='tight')
    plt.close()

def plot_error_distribution(data):
    """
    Gera um boxplot mostrando a distribuição dos erros para cada modelo
    """
    # Criar figura
    plt.figure(figsize=(15, 8))
    
    # Criar boxplot
    sns.boxplot(data=data.drop('source', axis=1))
    
    plt.title('Distribuição dos Erros por Modelo', pad=20, size=14)
    plt.xlabel('Modelos', labelpad=10)
    plt.ylabel('Erro', labelpad=10)
    
    # Rotacionar labels do eixo x
    plt.xticks(rotation=45, ha='right')
    
    plt.tight_layout()
    plt.savefig('error_distribution.png', dpi=300, bbox_inches='tight')
    plt.close()

def plot_correlation_matrix(data):
    """
    Gera um heatmap mostrando a correlação entre os erros dos diferentes modelos
    """
    # Calcular matriz de correlação
    corr_matrix = data.drop('source', axis=1).corr()
    
    # Criar figura
    plt.figure(figsize=(12, 10))
    
    # Criar heatmap
    sns.heatmap(corr_matrix, 
                annot=True, 
                fmt='.2f',
                cmap='coolwarm',
                center=0,
                square=True,
                cbar_kws={'label': 'Correlação'})
    
    plt.title('Matriz de Correlação entre os Erros dos Modelos', pad=20, size=14)
    
    # Rotacionar labels
    plt.xticks(rotation=45, ha='right')
    plt.yticks(rotation=0)
    
    plt.tight_layout()
    plt.savefig('correlation_matrix.png', dpi=300, bbox_inches='tight')
    plt.close()

# Gerar todas as visualizações
if __name__ == "__main__":
    # Ler os dados (assumindo que estejam em um arquivo chamado 'regression_Vazao_bbr_rmse.csv')
    try:
        data = pd.read_csv('regression_Vazao_bbr_rmse.csv')
        
        # Gerar todas as visualizações
        plot_model_comparison(data)
        plot_state_heatmap(data)
        plot_error_distribution(data)
        plot_correlation_matrix(data)
        
        print("Visualizações geradas com sucesso!")
        print("Arquivos gerados:")
        print("- model_comparison.png")
        print("- state_heatmap.png")
        print("- error_distribution.png")
        print("- correlation_matrix.png")
        
    except Exception as e:
        print(f"Erro ao gerar visualizações: {str(e)}")
```

```
import pandas as pd
import matplotlib.pyplot as plt

# Dados
data = {
    'RMSE': [
        0.222389,
        0.224132,
        0.227477,
        0.228550,
        0.231006,
        0.247009,
        0.267583,
        0.271097,
        0.293382,
        0.293403,
        0.343929,
        0.354753
    ]
}
index = [
    'CatBoostRegressor',
    'LGBMRegressor',
    'RandomForestRegressor',
    'XGBRegressor',
    'GradientBoostingRegressor',
    'KNeighborsRegressor',
    'AdaBoostRegressor',
    'PolynomialRegression',
    'ElasticNet',
    'LinearRegression',
    'MLPRegressor',
    'SVR'
]

# Cria o DataFrame original
df = pd.DataFrame(data, index=index)

# Ordena do maior para o menor RMSE
df_sorted = df.sort_values(by='RMSE', ascending=True)

# Transforma o índice em coluna e renomeia para "Modelos"
df2 = df_sorted.reset_index().rename(columns={'index': 'Modelos'})
# Agora df2 tem colunas: ["Modelos", "RMSE"] e está ordenado do maior para o menor

# Cria a figura e remove o eixo
fig, ax = plt.subplots(figsize=(3, 4))
ax.axis('off')

# Cria a tabela com df2
table = ax.table(
    cellText=df2.values,
    colLabels=df2.columns,
    cellLoc='center',
    loc='center'
)

# Ajuste de layout: fontes, larguras de coluna, etc.
table.auto_set_font_size(False)
table.set_fontsize(14)
table.auto_set_column_width(col=list(range(df2.shape[1])))

# Aumenta a altura de todas as linhas (inclusive cabeçalho)
for (row, col), cell in table.get_celld().items():
    cell.set_height(0.15)  # Ajuste conforme desejar

# Salva a figura
plt.savefig('tabela_rmse.png', bbox_inches='tight', dpi=300)
plt.show()
```

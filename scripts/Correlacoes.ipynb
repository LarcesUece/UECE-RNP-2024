```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose

caminho = r'..\resultados\prediction bi-LSTM interpolacao-linear - treated bbr esmond data ap-ba 07-03-2023_longest_interval.csv.csv'
df = pd.read_csv(caminho)

# Verificar se as colunas 'Test' e 'Prediction' estão presentes no dataframe
if 'Test' in df.columns and 'Prediction' in df.columns:
    # Calcular a correlação entre as duas colunas
    correlation = df['Test'].corr(df['Prediction'])
    print(f'Correlação entre "Test" e "Prediction": {correlation:.4f}')
    
    # Definir o índice como um datetime, se necessário (ajuste conforme seu dataset)
    # df['Datetime'] = pd.to_datetime(df['Datetime'])
    # df.set_index('Datetime', inplace=True)

    # Decompor a série temporal de Test e Prediction
    decomposition_test = seasonal_decompose(df['Test'], model='additive', period=1)  # Ajuste o período conforme necessário
    decomposition_prediction = seasonal_decompose(df['Prediction'], model='additive', period=1)

# Plotar os gráficos de decomposição comparando Test e Prediction
    plt.figure(figsize=(12, 10))

    # Observado
    plt.subplot(2, 2, 1)
    plt.plot(decomposition_test.observed, label='Test', color='blue')
    plt.plot(decomposition_prediction.observed, label='Prediction', color='red', linestyle='--')
    plt.title('Observado')
    plt.legend()

    # Tendência
    plt.subplot(2, 2, 2)
    plt.plot(decomposition_test.trend, label='Test', color='blue')
    plt.plot(decomposition_prediction.trend, label='Prediction', color='red', linestyle='--')
    plt.title('Tendência')
    plt.legend()

    # Sazonalidade
    plt.subplot(2, 2, 3)
    plt.plot(decomposition_test.seasonal, label='Test', color='blue')
    plt.plot(decomposition_prediction.seasonal, label='Prediction', color='red', linestyle='--')
    plt.title('Sazonalidade')
    plt.legend()

    # Resíduo
    plt.subplot(2, 2, 4)
    plt.plot(decomposition_test.resid, label='Test', color='blue')
    plt.plot(decomposition_prediction.resid, label='Prediction', color='red', linestyle='--')
    plt.title('Resíduo')
    plt.legend()

    # Exibir os gráficos de decomposição comparando Test e Prediction
    plt.tight_layout()
    plt.show()

else:
    print('As colunas "Test" e "Prediction" não foram encontradas no arquivo CSV.')
```

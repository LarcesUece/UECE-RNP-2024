```
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```
def plot_dtw_comparison(csv_path, output_path=None, figsize=(15, 10)):
    # Read the data
    df = pd.read_csv(csv_path)
    
    # Calculate mean values for each imputation method and DTW measure
    summary_df = df.groupby('imputation')[['dtw_seasonal', 'dtw_trend', 'dtw_observation']].mean().reset_index()
    
    # # Calculate standard deviation for error bars
    # std_df = df.groupby('imputation')[['dtw_seasonal', 'dtw_trend', 'dtw_observation']].std().reset_index()
    
    # Set up the plot style
    # plt.style.use('whitegrid')  # Use a valid style
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=figsize, height_ratios=[3, 1])
    
    # Create the grouped bar plot
    bar_width = 0.25
    index = np.arange(len(summary_df['imputation']))
    
    # Plot bars with standard deviation
    # bars1 = ax1.bar(index - bar_width, summary_df['dtw_seasonal'], bar_width, 
    #                 label='DTW Seasonal', color='#8884d8', 
    #                 yerr=std_df['dtw_seasonal'], capsize=5)
    # bars2 = ax1.bar(index, summary_df['dtw_trend'], bar_width,
    #                 label='DTW Trend', color='#82ca9d',
    #                 yerr=std_df['dtw_trend'], capsize=5)
    # bars3 = ax1.bar(index + bar_width, summary_df['dtw_observation'], bar_width,
    #                 label='DTW Observation', color='#ffc658',
    #                 yerr=std_df['dtw_observation'], capsize=5)

    bars1 = ax1.bar(index - bar_width, summary_df['dtw_seasonal'], bar_width, 
                    label='DTW Seasonal', color='#8884d8', capsize=5)
    bars2 = ax1.bar(index, summary_df['dtw_trend'], bar_width,
                    label='DTW Trend', color='#82ca9d', capsize=5)
    bars3 = ax1.bar(index + bar_width, summary_df['dtw_observation'], bar_width,
                    label='DTW Observation', color='#ffc658', capsize=5)
    
    # Customize the plot
    ax1.set_xlabel('Imputation Method')
    ax1.set_ylabel('DTW Value')
    ax1.set_title('DTW Analysis Comparison Across Imputation Methods')
    ax1.set_xticks(index)
    ax1.set_xticklabels(summary_df['imputation'], rotation=45, ha='right')
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    
    # Add value labels on top of bars
    def add_value_labels(bars):
        for bar in bars:
            height = bar.get_height()
            ax1.text(bar.get_x() + bar.get_width()/2., height,
                    f'{height:.3f}',
                    ha='center', va='bottom', rotation=0)
    
    add_value_labels(bars1)
    add_value_labels(bars2)
    add_value_labels(bars3)
    
    # Create a summary table
    summary_stats = pd.DataFrame({
        'Metric': ['Best Method (Seasonal)', 'Best Method (Trend)', 'Best Method (Observation)'],
        'Method': [
            summary_df.loc[summary_df['dtw_seasonal'].idxmin(), 'imputation'],
            summary_df.loc[summary_df['dtw_trend'].idxmin(), 'imputation'],
            summary_df.loc[summary_df['dtw_observation'].idxmin(), 'imputation']
        ],
        'Value': [
            summary_df['dtw_seasonal'].min(),
            summary_df['dtw_trend'].min(),
            summary_df['dtw_observation'].min()
        ]
    })
    
    # Plot the table
    ax2.axis('tight')
    ax2.axis('off')
    table = ax2.table(cellText=summary_stats.values,
                     colLabels=summary_stats.columns,
                     cellLoc='center',
                     loc='center',
                     colColours=['#f2f2f2']*3)
    table.auto_set_font_size(False)
    table.set_fontsize(9)
    table.scale(1.2, 1.5)
    
    # Adjust layout
    plt.tight_layout()
    
    # Save or show the plot
    if output_path:
        plt.savefig(output_path, bbox_inches='tight', dpi=300)
        print(f"Plot saved to {output_path}")
    else:
        plt.show()
    
    return summary_df, summary_stats

def sumary_dtw_statistics(summary_df, summary_stats):
    # Print numerical summary
    print("\nSummary Statistics:")
    print("==================")
    print("\nMean values for each imputation method:")
    print(summary_df.round(4))
    print("\nBest performing methods:")
    print(summary_stats)

def plot_dtw_models_comparison(csv_path, output_path=None, figsize=(20, 12)):
    # Read the data
    df = pd.read_csv(csv_path)
    
    # Create figure with subplots
    # plt.style.use('seaborn')
    fig = plt.figure(figsize=figsize)
    gs = fig.add_gridspec(3, 2, height_ratios=[3, 3, 1])
    
    # 1. Model Comparison Plot (left)
    ax1 = fig.add_subplot(gs[0, 0])
    model_summary = df.groupby('model')[['dtw_seasonal', 'dtw_trend', 'dtw_observation']].mean()
    # model_std = df.groupby('model')[['dtw_seasonal', 'dtw_trend', 'dtw_observation']].std()
    
    # Plot model comparison
    bar_width = 0.25
    index = np.arange(len(model_summary.index))
    
    # bars1 = ax1.bar(index - bar_width, model_summary['dtw_seasonal'], bar_width,
    #                 label='DTW Seasonal', color='#8884d8',
    #                 yerr=model_std['dtw_seasonal'], capsize=5)
    # bars2 = ax1.bar(index, model_summary['dtw_trend'], bar_width,
    #                 label='DTW Trend', color='#82ca9d',
    #                 yerr=model_std['dtw_trend'], capsize=5)
    # bars3 = ax1.bar(index + bar_width, model_summary['dtw_observation'], bar_width,
    #                 label='DTW Observation', color='#ffc658',
    #                 yerr=model_std['dtw_observation'], capsize=5)
    
    bars1 = ax1.bar(index - bar_width, model_summary['dtw_seasonal'], bar_width,
                    label='DTW Seasonal', color='#8884d8', capsize=5)
    bars2 = ax1.bar(index, model_summary['dtw_trend'], bar_width,
                    label='DTW Trend', color='#82ca9d', capsize=5)
    bars3 = ax1.bar(index + bar_width, model_summary['dtw_observation'], bar_width,
                    label='DTW Observation', color='#ffc658', capsize=5)
    
    ax1.set_title('DTW Analysis by Model Type')
    ax1.set_xticks(index)
    ax1.set_xticklabels(model_summary.index)
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    
    # 2. Imputation Method Comparison Plot (right)
    ax2 = fig.add_subplot(gs[0, 1])
    imp_summary = df.groupby('imputation')[['dtw_seasonal', 'dtw_trend', 'dtw_observation']].mean()
    # imp_std = df.groupby('imputation')[['dtw_seasonal', 'dtw_trend', 'dtw_observation']].std()
    
    # Plot imputation comparison
    index = np.arange(len(imp_summary.index))
    
    # bars4 = ax2.bar(index - bar_width, imp_summary['dtw_seasonal'], bar_width,
    #                 label='DTW Seasonal', color='#8884d8',
    #                 yerr=imp_std['dtw_seasonal'], capsize=5)
    # bars5 = ax2.bar(index, imp_summary['dtw_trend'], bar_width,
    #                 label='DTW Trend', color='#82ca9d',
    #                 yerr=imp_std['dtw_trend'], capsize=5)
    # bars6 = ax2.bar(index + bar_width, imp_summary['dtw_observation'], bar_width,
    #                 label='DTW Observation', color='#ffc658',
    #                 yerr=imp_std['dtw_observation'], capsize=5)

    bars4 = ax2.bar(index - bar_width, imp_summary['dtw_seasonal'], bar_width,
                    label='DTW Seasonal', color='#8884d8', capsize=5)
    bars5 = ax2.bar(index, imp_summary['dtw_trend'], bar_width,
                    label='DTW Trend', color='#82ca9d', capsize=5)
    bars6 = ax2.bar(index + bar_width, imp_summary['dtw_observation'], bar_width,
                    label='DTW Observation', color='#ffc658', capsize=5)
    
    ax2.set_title('DTW Analysis by Imputation Method')
    ax2.set_xticks(index)
    ax2.set_xticklabels(imp_summary.index, rotation=45, ha='right')
    ax2.legend()
    ax2.grid(True, alpha=0.3)
    
    # 3. Heatmap of Model-Imputation Combinations (bottom)
    ax3 = fig.add_subplot(gs[1, :])
    pivot_data = df.pivot_table(
        values='dtw_observation',  # Using observation as main metric
        index='model',
        columns='imputation',
        aggfunc='mean'
    )
    
    sns.heatmap(pivot_data, annot=True, fmt='.3f', cmap='YlOrRd', ax=ax3)
    ax3.set_title('DTW Observation Values for Model-Imputation Combinations')
    plt.xticks(rotation=45, ha='right')
    
    # 4. Summary Statistics Table
    ax4 = fig.add_subplot(gs[2, :])
    ax4.axis('tight')
    ax4.axis('off')
    
    # Prepare summary statistics
    best_combinations = pd.DataFrame({
        'Metric': ['Best Overall Combination', 'Best Model', 'Best Imputation Method'],
        'Configuration': [
            f"{df.loc[df['dtw_observation'].idxmin(), 'model']} + {df.loc[df['dtw_observation'].idxmin(), 'imputation']}",
            model_summary['dtw_observation'].idxmin(),
            imp_summary['dtw_observation'].idxmin()
        ],
        'DTW Observation': [
            df['dtw_observation'].min(),
            model_summary['dtw_observation'].min(),
            imp_summary['dtw_observation'].min()
        ]
    })
    
    table = ax4.table(cellText=best_combinations.values,
                     colLabels=best_combinations.columns,
                     cellLoc='center',
                     loc='center',
                     colColours=['#f2f2f2']*3)
    table.auto_set_font_size(False)
    table.set_fontsize(9)
    table.scale(1.2, 1.5)
    
    plt.tight_layout()
    
    # Save or show the plot
    if output_path:
        plt.savefig(output_path, bbox_inches='tight', dpi=300)
        print(f"Plot saved to {output_path}")
    else:
        plt.show()
    
    return model_summary, imp_summary, best_combinations

def generate_analysis_report(csv_path, output_path=None):
    """
    Generate a complete analysis report with statistical tests.
    """
    df = pd.read_csv(csv_path)
    
    # Perform statistical analysis
    model_stats = df.groupby('model').agg({
        'dtw_seasonal': ['mean', 'std', 'min', 'max'],
        'dtw_trend': ['mean', 'std', 'min', 'max'],
        'dtw_observation': ['mean', 'std', 'min', 'max']
    })
    
    imputation_stats = df.groupby('imputation').agg({
        'dtw_seasonal': ['mean', 'std', 'min', 'max'],
        'dtw_trend': ['mean', 'std', 'min', 'max'],
        'dtw_observation': ['mean', 'std', 'min', 'max']
    })
    
    # Create the report
    report = f"""
DTW Analysis Report
==================

Model Performance Summary
------------------------
{model_stats.round(4).to_string()}

Imputation Method Performance Summary
-----------------------------------
{imputation_stats.round(4).to_string()}
    """
    
    if output_path:
        with open(output_path, 'w') as f:
            f.write(report)
        print(f"Report saved to {output_path}")
    
    return report
```

```
# Example usage of the function
csv_path = "../results/DTW_imputation_original.csv"  # Replace with your CSV file path
output_path ="../results/dtw_analysis_imputation.png"  # Optional: path to save the figure
    
summary_df, summary_stats = plot_dtw_comparison(csv_path, output_path)
sumary_dtw_statistics(summary_df, summary_stats)
```

```
# melhores resultados de svd de dtw
def best_svd_dtw(path, index_list):
    df = pd.read_csv(path)
    df_bbr_filtered = df[df['tcp'] == 'bbr']
    pivot_df = df_bbr_filtered.pivot_table(
        index=index_list, 
        columns='imputation', 
        values=['dtw_seasonal', 'dtw_trend', 'dtw_observation']
    )
    imputations = ['interpolacao-linear', 'media-movel', 'mediana-movel', 'knn']
    conditions = [
        (pivot_df['dtw_observation', 'svd'] < pivot_df['dtw_observation', imp]) &
        (pivot_df['dtw_trend', 'svd'] < pivot_df['dtw_trend', imp]) &
        (pivot_df['dtw_seasonal', 'svd'] < pivot_df['dtw_seasonal', imp])
        for imp in imputations
    ]
    final_condition = conditions[0]
    for cond in conditions[1:]:
        final_condition &= cond

    best_svd = pivot_df[final_condition]
    return best_svd

original_imputation_dtw = best_svd_dtw('../results/DTW_imputation_original.csv', ['link', 'tcp'])
original_prediction_dtw  = best_svd_dtw('../results/DTW_LSTM_GRU.csv', ['link', 'tcp', 'model'])
print(original_prediction_dtw.index, original_imputation_dtw.index)
```

```
# verificando os melhores valores de svd (rmse) na predição
#path_predicao_rmse = '../graficos/predicoes/round_2/evaluation_rmse_mae_updated.csv'
def best_svd_prediction(path):
    df = pd.read_csv(path)
    pivot_df = df.pivot_table(index=['link', 'tcp', 'modelo'], columns='imputacao', values='RMSE')
    best_svd = pivot_df[
        (pivot_df['svd'] < pivot_df['interpolacao-linear']) &
        (pivot_df['svd'] < pivot_df['media-movel']) &
        (pivot_df['svd'] < pivot_df['mediana-movel']) &
        (pivot_df['svd'] < pivot_df['knn'])
    ]
    return best_svd
best_svd_predic = best_svd_prediction('../graficos/predicoes/round_2/evaluation_rmse_mae_updated.csv')

best_svd_predic.plot(kind='barh', figsize=(10, 15))
plt.title('Comparação de RMSE prediçãp', fontsize=14)
plt.xlabel('Link, TCP e Modelo', fontsize=12)
plt.ylabel('RMSE', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.legend(title='Imputação', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

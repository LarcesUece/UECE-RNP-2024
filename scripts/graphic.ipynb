```
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.ticker import ScalarFormatter


regressao_por_link = {
    'sp-mt-cubic': {
        'Estrategia1': 141.99522798168442,
        'Estrategia2': 136.18757993266577,
        'Estrategia3': 134.24515647852656
    },
    'sp-mt-bbr': {
        'Estrategia1': 10.661451768099807,
        'Estrategia2': 10.534804158487719,
        'Estrategia3': 10.499332623521209
    }
}
#SP-MT


# Preparando os dados para o gráfico
links = list(regressao_por_link.keys())

# Criando um conjunto de todas as categorias presentes
categorias_cubic = list(regressao_por_link['sp-mt-cubic'].keys())
categorias_bbr = list(regressao_por_link['sp-mt-bbr'].keys())
todas_categorias = list(set(categorias_cubic + categorias_bbr))

# Coletando os valores e preenchendo com 0 para categorias ausentes em cada link
valores_sp_m_cubic = [regressao_por_link['sp-mt-cubic'].get(cat, 0) for cat in todas_categorias]
valores_sp_m_bbr = [regressao_por_link['sp-mt-bbr'].get(cat, 0) for cat in todas_categorias]

# Configuração do gráfico
x = np.arange(len(todas_categorias))  # posição das categorias no eixo X
largura = 0.2  # largura das barras

fig, ax = plt.subplots(figsize=(10, 6))

# Gráfico de barras para cada link
barras_cubic = ax.bar(x - largura/2, valores_sp_m_cubic, largura, label='cubic')
barras_bbr = ax.bar(x + largura/2, valores_sp_m_bbr, largura, label='bbr')

# Personalizando o gráfico
ax.set_xlabel('Estrategias')
ax.set_ylabel('RMSE em Mbits/s')
ax.set_title('Comparação de Estrategias de Regressão para link SP-MT')
ax.set_xticks(x)
ax.set_xticklabels(todas_categorias, rotation=45, ha='right')
ax.legend()

# Remover notação científica
ax.yaxis.set_major_formatter(ScalarFormatter(useOffset=False))
ax.ticklabel_format(style='plain', axis='y')

plt.tight_layout()
plt.show()
```

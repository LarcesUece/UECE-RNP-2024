```
import pandas as pd
import os

def calculate_vazao_correlation_by_source(folder_path):
    """
    Calculate the correlation between 'Vazao' and 'Vazao_bbr' columns 
    for datasets with the same source (state) across multiple files.
    
    Parameters:
    folder_path (str): Path to the folder containing the datasets
    
    Returns:
    dict: A dictionary with correlation results grouped by source state
    """
    # Dictionary to store datasets grouped by source
    source_datasets = {}
    
    # Iterate through all files in the folder
    for filename in os.listdir(folder_path):
        if filename.endswith('.csv'):
            # Extract source (state) from filename
            try:
                source = filename.split('-')[2]
            except IndexError:
                print(f"Could not extract source from filename: {filename}")
                continue
            
            # Construct full file path
            file_path = os.path.join(folder_path, filename)
            
            try:
                # Read the CSV file
                df = pd.read_csv(file_path)
                
                # Check if required columns exist
                if 'Vazao' in df.columns and 'Vazao_bbr' in df.columns:
                    # Group datasets by source
                    if source not in source_datasets:
                        source_datasets[source] = []
                    source_datasets[source].append(df)
                else:
                    print(f"Required columns not found in {filename}")
            
            except Exception as e:
                print(f"Error processing {filename}: {e}")
    
    # Calculate correlations for each source
    correlation_results = {}
    for source, datasets in source_datasets.items():
        # Combine all datasets for the source
        combined_df = pd.concat(datasets, ignore_index=True)
        
        # Calculate correlations
        pearson_corr = combined_df['Vazao'].corr(combined_df['Vazao_bbr'])
        spearman_corr = combined_df['Vazao'].corr(combined_df['Vazao_bbr'], method='spearman')
        
        correlation_results[source] = {
            'Pearson Correlation': pearson_corr,
            'Spearman Correlation': spearman_corr,
            'Total Files': len(datasets),
            'Total Samples': len(combined_df)
        }
    
    return correlation_results

# Example usage
folder_path = '../datasets/serie-multivariada'  # Replace with your actual folder path
results = calculate_vazao_correlation_by_source(folder_path)

# Print results
for source, correlations in results.items():
    print(f"\nSource (State): {source}")
    print(f"Number of Files: {correlations['Total Files']}")
    print(f"Total Samples: {correlations['Total Samples']}")
    print(f"Pearson Correlation: {correlations['Pearson Correlation']:.4f}")
    print(f"Spearman Correlation: {correlations['Spearman Correlation']:.4f}")
```

```
import pandas as pd
import os

def calculate_vazao_correlation_by_source_destination(folder_path):
    # Dictionary to store datasets grouped by state
    state_datasets = {}
    
    # Iterate through all files in the folder
    for filename in os.listdir(folder_path):
        if filename.endswith('.csv'):
            # Extract source and destination states from filename
            try:
                source_state = filename.split('-')[2]
                destination_state = filename.split('-')[3]
            except IndexError:
                print(f"Could not extract states from filename: {filename}")
                continue
            
            # Construct full file path
            file_path = os.path.join(folder_path, filename)
            
            try:
                # Read the CSV file
                df = pd.read_csv(file_path)
                
                # Check if required columns exist
                if 'Vazao' in df.columns and 'Vazao_bbr' in df.columns:
                    # Add to datasets for source and destination states
                    for state in [source_state, destination_state]:
                        if state not in state_datasets:
                            state_datasets[state] = []
                        state_datasets[state].append(df)
                else:
                    print(f"Required columns not found in {filename}")
            
            except Exception as e:
                print(f"Error processing {filename}: {e}")
    
    # Calculate correlations for each state
    correlation_results = {}
    for state, datasets in state_datasets.items():
        # Combine all datasets for the state
        combined_df = pd.concat(datasets, ignore_index=True)
        
        # Calculate correlations
        pearson_corr = combined_df['Vazao'].corr(combined_df['Vazao_bbr'])
        spearman_corr = combined_df['Vazao'].corr(combined_df['Vazao_bbr'], method='spearman')
        
        correlation_results[state] = {
            'Pearson Correlation': pearson_corr,
            'Spearman Correlation': spearman_corr,
            'Total Files': len(datasets),
            'Total Samples': len(combined_df),
            'Datasets': [f.split('-')[-1] for f in os.listdir(folder_path) if 
                         f.endswith('.csv') and 
                         (state in f.split('-')[2] or state in f.split('-')[3])]
        }
    
    return correlation_results

# Example usage
folder_path = '../datasets/serie-multivariada'  # Replace with your actual folder path
results = calculate_vazao_correlation_by_source_destination(folder_path)

# Print results
for state, correlations in results.items():
    print(f"\nState: {state}")
    print(f"Number of Files: {correlations['Total Files']}")
    print(f"Total Samples: {correlations['Total Samples']}")
    print(f"Pearson Correlation: {correlations['Pearson Correlation']:.4f}")
    print(f"Spearman Correlation: {correlations['Spearman Correlation']:.4f}")
    # print("Datasets Used:")
    # for dataset in correlations['Datasets']:
    #     print(f"  - {dataset}")
```

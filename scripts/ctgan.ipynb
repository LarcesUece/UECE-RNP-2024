```
import pandas as pd 
datasets_path = '../datasets/datasets-by-source'
dataset_file = f'{datasets_path}/df_ap.csv'
data = pd.read_csv(dataset_file)
```

```
data.drop(columns=['Unnamed: 0'], inplace=True)
data
```

```
categorical_features = ['Link_bottleneck', 'Hop_count']
```

```
from ctgan import CTGAN

ctgan = CTGAN(verbose=True)
ctgan.fit(data, categorical_features, epochs = 500)
```

```
samples = ctgan.sample(1000)
```

```
samples
```

```
from table_evaluator import TableEvaluator

print(data.shape, samples.shape)
table_evaluator =  TableEvaluator(data[:1000], samples, cat_cols= categorical_features)

table_evaluator.visual_evaluation()
```
